```python
import heapq

def uniform_cost_search(graph, start, goal):
    """
    Performs Uniform Cost Search on a weighted graph.
    graph: Dict where keys are nodes and values are lists of tuples: (neighbor, cost)
    start: Starting node
    goal: Target node
    """
    # Priority queue stores tuples of (cumulative_cost, current_node, path)
    pq = [(0, start, [start])]

    # Dictionary to keep track of the cheapest cost to reach a visited node
    visited = {}

    while pq:
        cost, current, path = heapq.heappop(pq)

        # If we reach the goal, return the total cost and the path
        if current == goal:
            return cost, path

        # If the node has already been visited with a lower or equal cost, skip it
        if current in visited and visited[current] <= cost:
            continue

        visited[current] = cost

        # Explore neighbors
        for neighbor, weight in graph.get(current, []):
            new_cost = cost + weight

            if neighbor not in visited or new_cost < visited[neighbor]:
                heapq.heappush(pq, (new_cost, neighbor, path + [neighbor]))

    return None, None  # Return None if no path exists


# --- Example Usage ---
if __name__ == "__main__":
    # Graph representation: Node -> [(Neighbor, Cost), ...]
    weighted_graph = {
        'A': [('B', 1), ('C', 4)],
        'B': [('A', 1), ('C', 2), ('D', 5)],
        'C': [('A', 4), ('B', 2), ('D', 1)],
        'D': [('B', 5), ('C', 1)]
    }

    start_node = 'A'
    goal_node = 'D'

    total_cost, path = uniform_cost_search(weighted_graph, start_node, goal_node)

    if path:
        print(f"Shortest path found with cost {total_cost}:")
        print(" -> ".join(path))
    else:
        print("No path found.")
```
